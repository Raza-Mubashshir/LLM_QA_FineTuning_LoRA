```python
# ============================================================
# COMPLETE GOOGLE COLAB: Q&A FINE-TUNING WITH LoRA
# ============================================================

# -------------------------------
# 1. INSTALL REQUIRED LIBRARIES
# -------------------------------
!pip install -q -U transformers datasets peft accelerate bitsandbytes

# -------------------------------
# 2. IMPORT LIBRARIES
# -------------------------------
import os
import json
import torch

from google.colab import files
from datasets import Dataset

from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    BitsAndBytesConfig,
    TrainingArguments,
    Trainer,
    DataCollatorForLanguageModeling
)

from peft import (
    LoraConfig,
    get_peft_model,
    prepare_model_for_kbit_training
)

# -------------------------------
# 3. CHECK GPU
# -------------------------------
print("=" * 70)
print("GPU CHECK")
print("=" * 70)

if torch.cuda.is_available():
    print("GPU available:", torch.cuda.get_device_name(0))
else:
    print("WARNING: GPU is not available.")
    print("In Colab: Runtime -> Change runtime type -> T4 GPU")

# -------------------------------
# 4. UPLOAD JSON DATASET
# -------------------------------
print("\n" + "=" * 70)
print("UPLOAD YOUR JSON DATASET")
print("=" * 70)

uploaded = files.upload()

filename = next(iter(uploaded))

print("Uploaded file:", filename)

# -------------------------------
# 5. LOAD JSON
# -------------------------------
with open(filename, "r", encoding="utf-8") as f:
    json_data = json.load(f)

# Automatically handle common JSON formats
if isinstance(json_data, list):
    data = json_data

elif isinstance(json_data, dict):
    # Case: {"data": [...]}
    if "data" in json_data and isinstance(json_data["data"], list):
        data = json_data["data"]

    # Case: {"questions": [...]}
    elif "questions" in json_data and isinstance(json_data["questions"], list):
        data = json_data["questions"]

    # Case: single Q&A object
    elif "question" in json_data and "answer" in json_data:
        data = [json_data]

    else:
        raise ValueError(
            "Could not find Q&A data. Your JSON should contain "
            "'question' and 'answer' fields."
        )

else:
    raise ValueError("Unsupported JSON format.")

print("Number of Q&A examples:", len(data))

# -------------------------------
# 6. CHECK DATA
# -------------------------------
print("\nFirst raw example:")
print(data[0])

if "question" not in data[0] or "answer" not in data[0]:
    raise ValueError(
        "Each example must contain 'question' and 'answer' fields."
    )

# -------------------------------
# 7. CLEAN DATA
# -------------------------------
clean_data = []

for item in data:

    question = str(item["question"]).strip()
    answer = str(item["answer"]).strip()

    if question and answer:
        clean_data.append({
            "question": question,
            "answer": answer
        })

print("\nValid Q&A examples:", len(clean_data))

# -------------------------------
# 8. CREATE DATASET
# -------------------------------
dataset = Dataset.from_list(clean_data)

# -------------------------------
# 9. TRAIN / TEST SPLIT
# -------------------------------
if len(dataset) >= 5:

    split = dataset.train_test_split(
        test_size=0.2,
        seed=42
    )

    train_dataset = split["train"]
    eval_dataset = split["test"]

else:

    print("\nWARNING: Dataset is very small.")
    print("Using all examples for training.")

    train_dataset = dataset
    eval_dataset = dataset

print("\nTraining examples:", len(train_dataset))
print("Evaluation examples:", len(eval_dataset))

# -------------------------------
# 10. MODEL
# -------------------------------
MODEL_NAME = "Qwen/Qwen2.5-0.5B-Instruct"

print("\n" + "=" * 70)
print("LOADING MODEL")
print("=" * 70)

tokenizer = AutoTokenizer.from_pretrained(
    MODEL_NAME,
    trust_remote_code=True
)

if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

# -------------------------------
# 11. FORMAT Q&A
# -------------------------------
def format_example(example):

    text = (
        "### Question:\n"
        + example["question"]
        + "\n\n"
        + "### Answer:\n"
        + example["answer"]
        + tokenizer.eos_token
    )

    return {"text": text}


train_dataset = train_dataset.map(format_example)
eval_dataset = eval_dataset.map(format_example)

print("\nFormatted example:")
print(train_dataset[0]["text"])

# -------------------------------
# 12. TOKENIZATION
# -------------------------------
MAX_LENGTH = 512

def tokenize_function(examples):

    return tokenizer(
        examples["text"],
        truncation=True,
        max_length=MAX_LENGTH,
        padding="max_length"
    )

tokenized_train = train_dataset.map(
    tokenize_function,
    batched=True,
    remove_columns=train_dataset.column_names
)

tokenized_eval = eval_dataset.map(
    tokenize_function,
    batched=True,
    remove_columns=eval_dataset.column_names
)

print("\nTokenization completed.")

# -------------------------------
# 13. LOAD MODEL IN 4-BIT
# -------------------------------
print("\n" + "=" * 70)
print("LOADING 4-BIT MODEL")
print("=" * 70)

bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.float16,
    bnb_4bit_use_double_quant=True
)

model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    quantization_config=bnb_config,
    device_map="auto",
    trust_remote_code=True
)

model.config.use_cache = False

# -------------------------------
# 14. APPLY LoRA
# -------------------------------
print("\n" + "=" * 70)
print("APPLYING LoRA")
print("=" * 70)

model = prepare_model_for_kbit_training(model)

lora_config = LoraConfig(
    r=8,
    lora_alpha=16,
    target_modules=[
        "q_proj",
        "k_proj",
        "v_proj",
        "o_proj"
    ],
    lora_dropout=0.05,
    bias="none",
    task_type="CAUSAL_LM"
)

model = get_peft_model(
    model,
    lora_config
)

model.print_trainable_parameters()

# -------------------------------
# 15. TRAINING CONFIGURATION
# -------------------------------
training_args = TrainingArguments(
    output_dir="./lora_output",

    num_train_epochs=3,

    per_device_train_batch_size=2,
    per_device_eval_batch_size=2,

    gradient_accumulation_steps=4,

    learning_rate=2e-4,

    fp16=torch.cuda.is_available(),

    logging_steps=5,

    eval_strategy="epoch",
    save_strategy="epoch",

    save_total_limit=2,

    report_to="none",

    remove_unused_columns=False
)

# -------------------------------
# 16. DATA COLLATOR
# -------------------------------
data_collator = DataCollatorForLanguageModeling(
    tokenizer=tokenizer,
    mlm=False
)

# -------------------------------
# 17. TRAINER
# -------------------------------
trainer = Trainer(
    model=model,

    args=training_args,

    train_dataset=tokenized_train,

    eval_dataset=tokenized_eval,

    data_collator=data_collator
)

# -------------------------------
# 18. START FINE-TUNING
# -------------------------------
print("\n" + "=" * 70)
print("STARTING LoRA FINE-TUNING")
print("=" * 70)

trainer.train()

print("\nFine-tuning completed!")

# -------------------------------
# 19. SAVE LoRA MODEL
# -------------------------------
OUTPUT_DIR = "./fine_tuned_lora"

model.save_pretrained(OUTPUT_DIR)
tokenizer.save_pretrained(OUTPUT_DIR)

print("\nLoRA adapter saved to:", OUTPUT_DIR)

# -------------------------------
# 20. LOAD ORIGINAL MODEL
# -------------------------------
print("\n" + "=" * 70)
print("LOADING ORIGINAL MODEL FOR COMPARISON")
print("=" * 70)

original_model = AutoModelForCausalLM.from_pretrained(
    MODEL_NAME,
    quantization_config=bnb_config,
    device_map="auto",
    trust_remote_code=True
)

original_model.eval()

model.eval()

# -------------------------------
# 21. GENERATION FUNCTION
# -------------------------------
def generate_answer(model, question):

    prompt = (
        "### Question:\n"
        + question
        + "\n\n"
        + "### Answer:\n"
    )

    inputs = tokenizer(
        prompt,
        return_tensors="pt"
    )

    # Move tensors to model device
    inputs = {
        key: value.to(model.device)
        for key, value in inputs.items()
    }

    with torch.no_grad():

        outputs = model.generate(
            **inputs,

            max_new_tokens=100,

            do_sample=False,

            pad_token_id=tokenizer.pad_token_id,

            eos_token_id=tokenizer.eos_token_id
        )

    # Only decode newly generated tokens
    input_length = inputs["input_ids"].shape[1]

    generated_tokens = outputs[0][input_length:]

    answer = tokenizer.decode(
        generated_tokens,
        skip_special_tokens=True
    )

    return answer.strip()

# -------------------------------
# 22. NEW TEST QUESTIONS
# -------------------------------
# IMPORTANT:
# These should preferably NOT appear in your training dataset.

test_questions = [
    "What is machine learning?",
    "What is artificial intelligence?",
    "What is deep learning?",
    "What is LoRA?",
    "How does a neural network learn?"
]

# -------------------------------
# 23. COMPARE MODELS
# -------------------------------
print("\n" + "=" * 70)
print("ORIGINAL MODEL VS FINE-TUNED MODEL")
print("=" * 70)

results = []

for question in test_questions:

    print("\n" + "-" * 70)
    print("QUESTION:")
    print(question)

    # Original
    original_answer = generate_answer(
        original_model,
        question
    )

    # Fine-tuned
    finetuned_answer = generate_answer(
        model,
        question
    )

    print("\nORIGINAL MODEL:")
    print(original_answer)

    print("\nFINE-TUNED MODEL:")
    print(finetuned_answer)

    results.append({
        "question": question,
        "original_answer": original_answer,
        "finetuned_answer": finetuned_answer
    })

# -------------------------------
# 24. SAVE COMPARISON RESULTS
# -------------------------------
with open(
    "comparison_results.json",
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        results,
        f,
        indent=4,
        ensure_ascii=False
    )

print("\n" + "=" * 70)
print("COMPARISON SAVED")
print("=" * 70)

print("File: comparison_results.json")

# -------------------------------
# 25. DOWNLOAD RESULTS
# -------------------------------
files.download("comparison_results.json")

print("\nDONE!")
```
